# **Ice cream factory example**

H0 : mu <= 0.3

Ha : mu > 0.3

```python
import pandas as pd
import numpy as np
from scipy import stats
data=pd.Series([0.593,0.142,0.329,0.691,0.231,0.793,0.591,0.392,0.418])
data
```

```python
stats.ttest_1samp(data,0.3)
# population mean=0.3, this method always gives you 2 tail tests
#index of t-value is 0 and p value is 1
```

```python
p=stats.ttest_1samp(data,0.3)[1]
p
```

```python
#as this one tail test p value so devide by 2 to get final p_value
p_value=p/2
p_value
# Null hypo. is mu<=0.3,
#alter hypo. is mu>0.3
# alpha=0.05
#p_value we got is 0.026, so p_value<alpha. so reject null hypothesis
#go for alternate hypothesis
# interpret It : as mean value mu>0.3 factory is causing problem
```

# **Comparing drugs**

H0 : control = treat

HA : control != treat

```python
# 2 sample 2 tail test
control=pd.Series([91,87,99,77,88,91])
treat=pd.Series([101,110,103,93,99,104])
print(control)
print(treat)
```

```python
stats.ttest_ind(control,treat)
#two tail independent
#p=p1+p2 we are conducting 2 sample 2 tailtest so nop need to devide p by 2
#p=0.006, alpha=0.05, p<alpha so reject the null hypothesis
#interpretetion : 2 responses are not equal so drug is working
```

## **Placement data example **

H0 : p1 - p2 =0

Ha : p1 - p2 != 0

```python
n1=247
p1=0.37

n2=308
p2=0.39
```

```python
sample1= np.random.binomial(1,p1,n1)
sample1 # 247 Student's data 0-not placed 1-placed
```

```python
sample2= np.random.binomial(1,p2,n2)
sample2 # 308 Student's data 0-not placed 1-placed
```

```python
len(sample1)
```

```python
len(sample2)
```

```python
import statsmodels.api as sm # 2 proportion test funn is available in statsmodels.api
p=sm.stats.ttest_ind(sample1,sample2)[1] #2sample 2 tail test
#p>0.05 so accept the null hypothesis
#interpretation: students are almost equally got placed in 2 states
print(p)
print(np.round(p,3))
```
