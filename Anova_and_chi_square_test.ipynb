```python
# load libraries
from sklearn import datasets

#Load iris dataset
iris=datasets.load_iris()
iris #output will be dictionary
```

```python
group1_setosa=df[df["target"]==0]["sepal width (cm)"]
group2_versicolor=df[df["target"]==1]["sepal width (cm)"]
group3_virginica=df[df["target"]==2]["sepal width (cm)"]
```

```python
group1_setosa.mean(),group2_versicolor.mean(),group3_virginica.mean()
# mean of sepal width of all 3 categories
```

```python
import seaborn as sns
sns.displot(data=df,
             x="sepal width (cm)",
             hue="target",
             fill=True,
             kind="kde",
             palette=['blue','orange','green']
             )
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,5))
plt.scatter(range(1,51),group1_setosa,label="setosa")
plt.scatter(range(1,51),group2_versicolor,label="versicolor")
plt.scatter(range(1,51),group3_virginica,label="virginica")

plt.legend(iris.target_names)
plt.show()
```

```python
# f_oneway - one way ANOVA
from scipy import stats
stats.f_oneway(group1_setosa,group2_versicolor,group3_virginica)
# p-value is 4.49*10^-17 which is approximately equal to 0. so p<alpha
# so reject the null hypothisis
#interpretetation:atleast one of them is different. i.e, setosa is different

```

```python
#another example of one way ANOVA
df1=pd.DataFrame(iris.data)
stats.f_oneway(df1.iloc[:,0],df1.iloc[:,1],df1.iloc[:,2],df1.iloc[:,3])
# p-value is 4.66*10^-159 which is approximately equal to 0. so p<alpha
# H0: MUs1 = MUsw = MUpw, Ha = atleast one of them is different
```

## **chi-square test**

```python
from scipy.stats import chi2_contingency
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#athlete and smoker..example
data=[[14,4],
      [0,10]]
data
```

```python
df=pd.DataFrame(data,
                index=["Athlete","Non Athlete"],
                columns=["Non-smoker","Smoker"])
df
```

```python
df.plot(kind="bar",stacked="False")
```

```python
chi2,p,dof,exp=chi2_contingency(data) # test statistic, p_value, df, expected frequency
#p-value 0.00038<0.05 so reject the null hypothesis
```

```python
chi2_contingency(data)
```

```python
chi2
```

```python
dof
```

```python
exp
```

```python
a=0.05
if p<a:
  print("reject the null hypothesis. smoking and being athlete are depend on each other")
else:
  print("accept the null hypothesis. smoking and being athlete are not depend on each other")
```
